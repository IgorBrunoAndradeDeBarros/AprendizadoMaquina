# 📊 Aula 2 - Pandas, Matplotlib e Seaborn - Aula Prática

> **Foco:** manipulação de dados tabulares, funções de agregação, leitura/escrita de Excel e visualização de gráficos.

Este notebook usa um dataset de **desempenho acadêmico** (120 alunos, 4 disciplinas) gerado internamente.  
Execute cada célula na ordem com **Shift + Enter**.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 110
np.random.seed(42)

print("Bibliotecas carregadas com sucesso.")
```

```python
# Gera o dataset e salva em Excel para usarmos na aula
n = 120
turmas = np.repeat(["A", "B", "C"], n // 3)
nomes_base = ["Ana","Bruno","Carla","Diego","Elena","Felipe","Gabriela","Hugo",
              "Isabela","João","Karen","Lucas","Mariana","Nicolas","Olivia",
              "Pedro","Queila","Rafael","Sara","Thiago","Ursula","Vitor","Wendy","Xavier","Yasmin","Zé"]
nomes = [f"{nomes_base[i % len(nomes_base)]} {chr(65 + i // len(nomes_base))}" for i in range(n)]

def notas(media, std, n):
    return np.clip(np.random.normal(media, std, n), 0, 10).round(1)

df_origem = pd.DataFrame({
    "Nome"         : nomes,
    "Turma"        : turmas,
    "Matematica"   : np.concatenate([notas(6.5,1.8,40), notas(7.0,1.5,40), notas(5.8,2.0,40)]),
    "Estatistica"  : np.concatenate([notas(6.0,1.9,40), notas(6.8,1.6,40), notas(6.2,2.1,40)]),
    "Programacao"  : np.concatenate([notas(7.2,1.4,40), notas(7.5,1.3,40), notas(6.0,1.8,40)]),
    "Ciencia_Dados": np.concatenate([notas(6.8,1.7,40), notas(7.1,1.5,40), notas(5.5,2.2,40)]),
    "Frequencia"   : np.clip(np.random.normal(82, 12, n), 0, 100).round(1),
})

df_origem.to_excel("alunos.xlsx", index=False)
print(f"Arquivo 'alunos.xlsx' criado — {df_origem.shape[0]} linhas × {df_origem.shape[1]} colunas")
```

## 1. Pandas — Leitura e Inspeção de Dados

### 1.1 Abrindo um arquivo Excel

`pd.read_excel()` lê qualquer `.xlsx` e devolve um **DataFrame** — a estrutura tabular central do Pandas.

| Parâmetro | O que faz |
|---|---|
| `sheet_name` | Nome ou índice da aba (padrão: primeira aba) |
| `usecols` | Seleciona apenas certas colunas |
| `skiprows` | Pula linhas de cabeçalho extra |
| `dtype` | Força tipo de uma coluna |

```python
df = pd.read_excel("alunos.xlsx")   # lê a primeira aba do arquivo

df.head()   # exibe as 5 primeiras linhas
```

### 1.2 Inspeção Rápida do Dataset

```python
print("--- shape (linhas, colunas) ---")
print(df.shape)

print("\n--- tipos de cada coluna ---")
print(df.dtypes)

print("\n--- valores nulos por coluna ---")
print(df.isnull().sum())
```

```python
# describe() resume as colunas numéricas em 8 estatísticas de uma vez
df.describe().round(2)
```

### 1.3 Seleção de Colunas e Filtros

```python
# -- Selecionar colunas ----------------------------------------------------
print("Uma coluna (retorna Series):")
print(df["Turma"].value_counts())   # contagem por turma

print("\nVárias colunas (retorna DataFrame):")
print(df[["Nome", "Turma", "Matematica"]].head(4).to_string(index=False))

# -- Filtros booleanos -----------------------------------------------------
print("\nAlunos com nota de Matemática acima de 9:")
alunos_destaque = df[df["Matematica"] > 9][["Nome", "Turma", "Matematica"]]
print(alunos_destaque.to_string(index=False))

# -- Múltiplas condições ---------------------------------------------------
print("\nTurma B com Frequência abaixo de 70%:")
baixa_freq = df[(df["Turma"] == "B") & (df["Frequencia"] < 70)]
print(baixa_freq[["Nome", "Frequencia"]].to_string(index=False))
```

### 1.4 Criando Novas Colunas

```python
disc = ["Matematica", "Estatistica", "Programacao", "Ciencia_Dados"]

# Média das 4 disciplinas (axis=1 → calcula linha a linha)
df["Media"] = df[disc].mean(axis=1).round(2)

# Situação: aprovado se média >= 5 E frequência >= 75
df["Aprovado"] = (df["Media"] >= 5) & (df["Frequencia"] >= 75)

# Classificação em texto usando np.select (equivale a IF-ELIF-ELSE)
condicoes   = [df["Media"] >= 8, df["Media"] >= 6, df["Media"] >= 5]
rotulos     = ["Excelente", "Bom", "Regular"]
df["Conceito"] = np.select(condicoes, rotulos, default="Insuficiente")

df[["Nome", "Turma", "Media", "Aprovado", "Conceito"]].head(8)
```

### 1.5 Ordenação e Ranking

```python
# Top 5 alunos por média
print("Top 5 — Melhores Médias:")
print(df.nlargest(5, "Media")[["Nome", "Turma", "Media", "Conceito"]].to_string(index=False))

# Pior desempenho
print("\nBottom 5 — Menores Médias:")
print(df.nsmallest(5, "Media")[["Nome", "Turma", "Media", "Aprovado"]].to_string(index=False))

# Ranking dentro de cada turma (dense = sem pulos de posição)
df["Rank_Turma"] = df.groupby("Turma")["Media"].rank(method="dense", ascending=False).astype(int)
print("\nRanking por turma (primeiros de cada):")
print(df[df["Rank_Turma"] == 1][["Nome", "Turma", "Media", "Rank_Turma"]].to_string(index=False))
```

## 2. Funções de Agregação

Agregação = **reduzir** um conjunto de linhas a um único valor (ou conjunto de valores) por grupo.

### 2.1 `groupby` + funções diretas

As funções mais comuns aplicadas após `groupby`:

| Função | O que retorna |
|---|---|
| `.mean()` | Média |
| `.sum()` | Soma |
| `.count()` | Contagem de valores não-nulos |
| `.min()` / `.max()` | Mínimo / Máximo |
| `.std()` | Desvio-padrão |
| `.median()` | Mediana |

```python
# Média de todas as colunas numéricas por turma
print("Média por turma:")
print(df.groupby("Turma")[disc + ["Media", "Frequencia"]].mean().round(2))

print("\nContagem de alunos por turma:")
print(df.groupby("Turma")["Nome"].count())

print("\nMáxima e mínima média por turma:")
print(df.groupby("Turma")["Media"].agg(["max", "min"]).round(2))
```

### 2.2 `agg` com Múltiplas Funções ao Mesmo Tempo

`agg` permite aplicar **funções diferentes em colunas diferentes** em uma única chamada.

```python
resumo = df.groupby("Turma").agg(
    Alunos        = ("Nome",        "count"),
    Media_Geral   = ("Media",       "mean"),
    Media_Max     = ("Media",       "max"),
    Media_Min     = ("Media",       "min"),
    Desvio_Padrao = ("Media",       "std"),
    Mediana       = ("Media",       "median"),
    Freq_Media    = ("Frequencia",  "mean"),
    Aprovados     = ("Aprovado",    "sum"),   # True conta como 1
).round(2)

print("Resumo por turma:")
resumo
```

### 2.3 `pivot_table` — Tabela Dinâmica

`pivot_table` é equivalente à tabela dinâmica do Excel: cruza duas variáveis categóricas e agrega uma variável numérica.

```python
# Média por Turma × Conceito
tabela = pd.pivot_table(
    df,
    values  = "Media",       # valor a agregar
    index   = "Turma",       # linhas
    columns = "Conceito",    # colunas
    aggfunc = "count",       # contagem de alunos em cada célula
    fill_value = 0
)

print("Distribuição de alunos por Turma × Conceito:")
tabela
```

### 2.4 `value_counts` e `crosstab`

```python
# value_counts: distribuição de uma coluna categórica
print("Distribuição por Conceito:")
print(df["Conceito"].value_counts())

print("\nDistribuição relativa (%):")
print((df["Conceito"].value_counts(normalize=True) * 100).round(1))

# crosstab: tabela de frequência cruzada entre duas colunas categóricas
print("\nCrosstab Turma × Aprovado:")
print(pd.crosstab(df["Turma"], df["Aprovado"],
                  rownames=["Turma"], colnames=["Aprovado"],
                  margins=True, margins_name="Total"))
```

### 2.5 Salvando Resultados em Excel

`pd.ExcelWriter` abre uma sessão de escrita e permite criar **múltiplas abas** no mesmo arquivo.

```python
with pd.ExcelWriter("resultado_aula.xlsx", engine="openpyxl") as writer:
    # Aba 1: dados completos com colunas derivadas
    df.to_excel(writer, sheet_name="Dados_Completos", index=False)

    # Aba 2: resumo agregado por turma
    resumo.to_excel(writer, sheet_name="Resumo_Turmas")

    # Aba 3: tabela dinâmica Turma × Conceito
    tabela.to_excel(writer, sheet_name="Pivot_Conceito")

    # Aba 4: top 10 alunos
    df.nlargest(10, "Media")[["Nome", "Turma", "Media", "Conceito"]].to_excel(
        writer, sheet_name="Top10", index=False
    )

print("'resultado_aula.xlsx' salvo com 4 abas:"
      "\n  • Dados_Completos  • Resumo_Turmas  • Pivot_Conceito  • Top10")
```

## 3. Matplotlib — Plotagem

Matplotlib oferece controle total sobre os gráficos. O padrão moderno é usar `fig, ax = plt.subplots()`.

> **Anatomia rápida:** `Figure` (tela) → `Axes` (área do gráfico) → `Artists` (linhas, barras, textos)

### 3.1 Gráfico de Barras (`bar`)

```python
medias_turma = df.groupby("Turma")["Media"].mean()

fig, ax = plt.subplots(figsize=(7, 4))

bars = ax.bar(medias_turma.index, medias_turma.values,
              color=["#2196F3", "#4CAF50", "#FF9800"],
              edgecolor="white", linewidth=0.8)

# Rótulo de valor em cima de cada barra
ax.bar_label(bars, fmt="%.2f", padding=4, fontsize=10)

ax.set_title("Média Final por Turma", fontsize=13, fontweight="bold")
ax.set_xlabel("Turma")
ax.set_ylabel("Média")
ax.set_ylim(0, 10)
ax.axhline(5, color="red", linestyle="--", linewidth=1, alpha=0.6, label="Mínimo (5.0)")
ax.legend()

plt.tight_layout()
plt.show()
```

### 3.2 Histograma (`hist`)

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.hist(df["Media"], bins=20, color="#5C6BC0", edgecolor="white", linewidth=0.8)

ax.axvline(df["Media"].mean(),   color="#E53935", linestyle="--",
           linewidth=1.8, label=f"Média: {df['Media'].mean():.2f}")
ax.axvline(df["Media"].median(), color="#43A047", linestyle="-.",
           linewidth=1.8, label=f"Mediana: {df['Media'].median():.2f}")
ax.axvline(5.0, color="orange", linestyle=":", linewidth=1.4, label="Mínimo: 5.0")

ax.set_title("Distribuição das Médias Finais", fontsize=13, fontweight="bold")
ax.set_xlabel("Média Final")
ax.set_ylabel("Número de Alunos")
ax.legend()

plt.tight_layout()
plt.show()
```

### 3.3 Scatter Plot (`scatter`)

```python
cores = df["Turma"].map({"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"})

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["Frequencia"], df["Media"], c=cores, alpha=0.65, s=50)

# Linha de tendência
z = np.polyfit(df["Frequencia"], df["Media"], 1)
x_lin = np.linspace(df["Frequencia"].min(), df["Frequencia"].max(), 200)
ax.plot(x_lin, np.poly1d(z)(x_lin), "k--", linewidth=1.3, alpha=0.5, label="Tendência")

ax.axhline(5,  color="red",  linestyle=":", linewidth=1, alpha=0.6)
ax.axvline(75, color="gray", linestyle="--", linewidth=1, alpha=0.5, label="Freq. mín. 75%")

ax.set_title("Frequência × Média Final", fontsize=13, fontweight="bold")
ax.set_xlabel("Frequência (%)")
ax.set_ylabel("Média Final")
ax.legend()

plt.tight_layout()
plt.show()

print(f"Correlação Frequência × Média: {df['Frequencia'].corr(df['Media']):.3f}")
```

### 3.4 Subplots — Painel com 4 Gráficos

`plt.subplots(linhas, colunas)` cria uma grade de eixos. Cada `ax` é acessado como `axes[i, j]`.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Visão Geral do Desempenho", fontsize=14, fontweight="bold")

pal = {"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"}

# [0,0] Barras — média por turma
medias = df.groupby("Turma")["Media"].mean()
axes[0, 0].bar(medias.index, medias.values, color=pal.values(), edgecolor="white")
axes[0, 0].set_title("Média por Turma"); axes[0, 0].set_ylim(0, 10)
axes[0, 0].axhline(5, color="red", linestyle=":", linewidth=1)

# [0,1] Histograma — distribuição de médias
axes[0, 1].hist(df["Media"], bins=18, color="#5C6BC0", edgecolor="white")
axes[0, 1].axvline(df["Media"].mean(), color="red", linestyle="--", linewidth=1.5)
axes[0, 1].set_title("Distribuição das Médias")

# [1,0] Pizza — aprovados vs reprovados
contagem = df["Aprovado"].value_counts()
axes[1, 0].pie(
    contagem, labels=["Aprovado", "Reprovado"],
    colors=["#4CAF50", "#EF5350"], autopct="%1.1f%%",
    wedgeprops={"edgecolor": "white"}
)
axes[1, 0].set_title("Aprovação Geral")

# [1,1] Scatter — frequência × média
for turma, cor in pal.items():
    m = df["Turma"] == turma
    axes[1, 1].scatter(df.loc[m, "Frequencia"], df.loc[m, "Media"],
                       color=cor, alpha=0.55, s=30, label=turma)
axes[1, 1].set_title("Frequência × Média")
axes[1, 1].set_xlabel("Frequência (%)")
axes[1, 1].legend(title="Turma", fontsize=8)

plt.tight_layout()
plt.show()
```

## 4. Seaborn — Gráficos Estatísticos

**Seaborn** é construído sobre Matplotlib e integra nativamente com DataFrames do Pandas.  
Ele cuida da estética automaticamente e adiciona intervalos de confiança e estimativas de densidade sem código extra.

### 4.1 `barplot` — Médias com Intervalo de Confiança

```python
# Formato "longo": uma linha por (aluno, disciplina)
df_longo = df.melt(
    id_vars=["Nome", "Turma"],
    value_vars=disc,
    var_name="Disciplina",
    value_name="Nota"
)

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(
    data=df_longo,
    x="Disciplina", y="Nota", hue="Turma",
    palette={"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"},
    capsize=0.07, errwidth=1.5, ax=ax
)
ax.axhline(5, color="red", linestyle=":", linewidth=1.2)
ax.set_title("Média ± IC 95% por Disciplina e Turma", fontsize=13, fontweight="bold")
ax.set_xticklabels([d.replace("_", " ") for d in disc])
ax.set_ylabel("Nota")
ax.legend(title="Turma")
plt.tight_layout()
plt.show()
```

### 4.2 `histplot` com KDE — Distribuição por Turma

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Histograma geral com curva de densidade (kde=True)
sns.histplot(df["Media"], bins=20, kde=True, color="#5C6BC0", ax=axes[0])
axes[0].set_title("Distribuição Geral da Média")
axes[0].set_xlabel("Média Final")

# KDE separado por turma (fill=True preenche sob a curva)
for turma, cor in {"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"}.items():
    sns.kdeplot(df[df["Turma"] == turma]["Media"],
                label=f"Turma {turma}", fill=True, alpha=0.25,
                color=cor, ax=axes[1])
axes[1].set_title("Densidade por Turma")
axes[1].set_xlabel("Média Final")
axes[1].legend()

plt.tight_layout()
plt.show()
```

### 4.3 `boxplot` — Dispersão das Notas por Disciplina e Turma

O boxplot mostra: **mediana** (linha central), **1º e 3º quartis** (caixa), **mínimo/máximo** (hastes) e **outliers** (pontos).

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Boxplot da média final por turma
sns.boxplot(data=df, x="Turma", y="Media",
            palette={"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"},
            linewidth=1.5, ax=axes[0])
axes[0].axhline(5, color="red", linestyle="--", linewidth=1, alpha=0.7)
axes[0].set_title("Média Final por Turma")
axes[0].set_ylabel("Média Final")

# Boxplot de cada disciplina (formato longo)
sns.boxplot(data=df_longo, x="Disciplina", y="Nota", hue="Turma",
            palette={"A": "#2196F3", "B": "#4CAF50", "C": "#FF9800"},
            linewidth=1.2, ax=axes[1])
axes[1].axhline(5, color="red", linestyle="--", linewidth=1, alpha=0.7)
axes[1].set_title("Notas por Disciplina e Turma")
axes[1].set_xticklabels([d.replace("_", " ") for d in disc], rotation=15)
axes[1].legend(title="Turma", fontsize=8)

plt.tight_layout()
plt.show()
```

### 4.4 `heatmap` — Mapa de Calor da Correlação

A matriz de correlação mostra o grau de relação linear entre cada par de variáveis numéricas (-1 a +1).

```python
corr = df[disc + ["Frequencia", "Media"]].corr()

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True, fmt=".2f",
    cmap="RdYlGn", vmin=-1, vmax=1,
    linewidths=0.5, square=True, ax=ax
)
ax.set_title("Matriz de Correlação", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()

print("\nColunas mais correlacionadas com a Média Final:")
print(corr["Media"].drop("Media").sort_values(ascending=False).to_string())
```

## 📌 Resumo Rápido

| Tarefa | Código |
|---|---|
| Abrir Excel | `pd.read_excel("arquivo.xlsx")` |
| Salvar Excel (múltiplas abas) | `pd.ExcelWriter` + `.to_excel(writer, sheet_name=...)` |
| Nova coluna calculada | `df["col"] = df["a"] + df["b"]` |
| Coluna condicional | `np.where(condição, valor_sim, valor_não)` |
| Agrupar e agregar | `df.groupby("col").agg(novo_nome=("col", "func"))` |
| Tabela dinâmica | `pd.pivot_table(df, values, index, columns, aggfunc)` |
| Distribuição de categorias | `df["col"].value_counts()` |
| Tabela cruzada | `pd.crosstab(df["a"], df["b"])` |
| Barras (matplotlib) | `ax.bar(x, y)` + `ax.bar_label(bars)` |
| Histograma (seaborn) | `sns.histplot(df["col"], kde=True)` |
| Boxplot (seaborn) | `sns.boxplot(data=df, x="cat", y="num")` |
| Heatmap de correlação | `sns.heatmap(df.corr(), annot=True, cmap="RdYlGn")` |
