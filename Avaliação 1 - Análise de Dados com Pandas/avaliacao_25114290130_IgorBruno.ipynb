## ⚙️ Setup - Não modifique estas células

```python
import numpy as np
import pandas as pd

pd.set_option("display.float_format", "{:.2f}".format)
pd.set_option("display.max_columns", 20)
np.random.seed(99)
```

```python
# Geração do dataset - execute uma vez para criar o arquivo Excel
categorias = ["Eletrônicos", "Alimentos", "Vestuário", "Esportes", "Casa"]
produtos = {
    "Eletrônicos": ["Notebook", "Smartphone", "Tablet", "Fone Bluetooth", "Smartwatch", "Carregador"],
    "Alimentos"  : ["Arroz 5kg", "Feijão 1kg", "Azeite 500ml", "Café 500g", "Macarrão 500g", "Leite 1L"],
    "Vestuário"  : ["Camiseta", "Calça Jeans", "Tênis", "Jaqueta", "Meias Kit", "Bermuda"],
    "Esportes"   : ["Bola de Futebol", "Raquete Tênis", "Luva Boxe", "Colchonete", "Halteres", "Corda Pular"],
    "Casa"       : ["Panela Antiaderente", "Travesseiro", "Toalha Banho", "Vassoura", "Porta Shampoo", "Lençol"],
}
vendedores = ["Ana Lima", "Carlos Souza", "Fernanda Rocha", "Marcos Alves", "Patrícia Costa"]
regioes    = ["Norte", "Sul", "Leste", "Oeste"]

n = 500
datas = pd.date_range("2024-01-01", "2024-12-31", periods=n)
cats  = np.random.choice(categorias, n)

registros = []
for i in range(n):
    cat     = cats[i]
    prod    = np.random.choice(produtos[cat])
    preco   = round(np.random.uniform(
                  5 if cat == "Alimentos" else 30,
                 (800 if cat == "Eletrônicos" else 150)
              ), 2)
    registros.append({
        "ID_Venda"       : 1000 + i,
        "Data"           : datas[i].date(),
        "Produto"        : prod,
        "Categoria"      : cat,
        "Vendedor"       : np.random.choice(vendedores),
        "Regiao"         : np.random.choice(regioes),
        "Qtd"            : int(np.random.randint(1, 11)),
        "Preco_Unitario" : preco,
        "Desconto_Pct"   : round(np.random.choice([0, 0, 0, 5, 5, 10, 15, 20]), 1),
    })

pd.DataFrame(registros).to_excel("vendas_supermercado.xlsx", index=False)
print("Arquivo 'vendas_supermercado.xlsx' gerado com", n, "registros.")
```

### Tarefa 2.1 - Subtotal e Desconto *(1 pt)*

| Coluna nova | Fórmula |
|---|---|
| `Subtotal` | `Qtd × Preco_Unitario` |
| `Desconto_R$` | `Subtotal × (Desconto_Pct / 100)` |
| `Total_Liquido` | `Subtotal − Desconto_R$` |

Exiba: `ID_Venda`, `Produto`, `Qtd`, `Preco_Unitario`, `Desconto_Pct`, `Subtotal`, `Desconto_R$`, `Total_Liquido`

```python
# SEU CÓDIGO AQUI - Tarefa 2.1

df["Subtotal"] = df["Qtd"] * df["Preco_Unitario"]
df["Desconto_R$"] = df["Subtotal"] * (df["Desconto_Pct"] / 100)
df["Total_Liquido"] = df["Subtotal"] - df["Desconto_R$"]

display(df[["ID_Venda","Produto","Qtd","Preco_Unitario","Desconto_Pct",
            "Subtotal","Desconto_R$","Total_Liquido"]].head(6))
```

### Tarefa 2.2 - Colunas de Data *(0,5 pt)*

A coluna `Data` é do tipo `object`. Converta-a para `datetime` e então extraia:

| Coluna nova | Conteúdo |
|---|---|
| `Mes` | Número do mês (1–12) |
| `Trimestre` | Trimestre (1–4) |
| `Nome_Mes` | Nome do mês em português (`Janeiro`, `Fevereiro`, …) |

> **Dica:** `pd.to_datetime()` para converter; `.dt.month`, `.dt.quarter` e `.dt.month_name(locale="pt_BR.UTF-8")` para extrair - ou crie um dicionário de mapeamento se o locale não estiver disponível.

Exiba: `ID_Venda`, `Data`, `Mes`, `Trimestre`, `Nome_Mes`

```python
# SEU CÓDIGO AQUI - Tarefa 2.2
df["Data"] = pd.to_datetime(df["Data"])
df["Mes"] = df["Data"].dt.month
df["Trimestre"] = df["Data"].dt.quarter

meses_pt = {1:"Janeiro",2:"Fevereiro",3:"Março",4:"Abril",5:"Maio",6:"Junho",
            7:"Julho",8:"Agosto",9:"Setembro",10:"Outubro",11:"Novembro",12:"Dezembro"}
df["Nome_Mes"] = df["Mes"].map(meses_pt)

display(df[["ID_Venda","Data","Mes","Trimestre","Nome_Mes"]].head(6))
```

### Tarefa 2.3 - Métricas Agregadas por Produto *(1 pt)*

Calcule, **por produto**, as seguintes métricas e junte-as de volta ao `df` como novas colunas:

| Coluna nova | Descrição |
|---|---|
| `Media_Total_Produto` | Média do `Total_Liquido` de todas as vendas daquele produto |
| `Soma_Total_Produto` | Soma total do `Total_Liquido` por produto |
| `Qtd_Vendas_Produto` | Número de vezes que o produto aparece no dataset |

> **Dica:** use `groupby` + `transform` para que o resultado fique alinhado linha a linha com o DataFrame original.

Exiba: `Produto`, `Total_Liquido`, `Media_Total_Produto`, `Soma_Total_Produto`, `Qtd_Vendas_Produto` das **10 primeiras linhas**, sem repetir produtos (`drop_duplicates("Produto")`).

```python
# SEU CÓDIGO AQUI - Tarefa 2.3
df["Media_Total_Produto"] = df.groupby("Produto")["Total_Liquido"].transform("mean")
df["Soma_Total_Produto"] = df.groupby("Produto")["Total_Liquido"].transform("sum")
df["Qtd_Vendas_Produto"] = df.groupby("Produto")["Total_Liquido"].transform("count")

display(df[["Produto","Total_Liquido","Media_Total_Produto",
            "Soma_Total_Produto","Qtd_Vendas_Produto"]].drop_duplicates("Produto").head(10))
```

### Tarefa 2.4 - Classificação do Ticket *(0,75 pt)*

Crie a coluna `Faixa_Ticket` com base no `Total_Liquido` de cada venda:

| Faixa | Condição |
|---|---|
| `"Baixo"` | Total_Liquido < 100 |
| `"Médio"` | 100 ≤ Total_Liquido < 500 |
| `"Alto"` | Total_Liquido ≥ 500 |

> Use `np.select` ou `pd.cut`.

Exiba a contagem de cada faixa com `value_counts()`.

```python
# SEU CÓDIGO AQUI - Tarefa 2.4
condicoes = [
    df["Total_Liquido"] < 100,
    (df["Total_Liquido"] >= 100) & (df["Total_Liquido"] < 500),
    df["Total_Liquido"] >= 500
]
valores = ["Baixo", "Médio", "Alto"]
df["Faixa_Ticket"] = np.select(condicoes, valores)

print(df["Faixa_Ticket"].value_counts())
```

### Tarefa 2.5 - Lucro Estimado e Comissão *(0,75 pt)*

Adicione:

| Coluna nova | Fórmula | Observação |
|---|---|---|
| `Lucro_Estimado` | `Total_Liquido × 0.30` | Margem de 30% sobre o valor líquido |
| `Comissao_Vendedor` | `Total_Liquido × 0.02` | 2% do valor líquido para o vendedor |

Em seguida, calcule e imprima na tela:
- O **lucro total estimado** de todas as vendas
- O **total de comissões** pagas
- A **comissão média por vendedor** (use `groupby("Vendedor")["Comissao_Vendedor"].sum()`)

```python
# SEU CÓDIGO AQUI - Tarefa 2.5
df["Lucro_Estimado"] = df["Total_Liquido"] * 0.30
df["Comissao_Vendedor"] = df["Total_Liquido"] * 0.02

print("Lucro total estimado: R$", round(df["Lucro_Estimado"].sum(), 2))
print("Total de comissões: R$", round(df["Comissao_Vendedor"].sum(), 2))
print("\nComissão total por vendedor:")
print(df.groupby("Vendedor")["Comissao_Vendedor"].sum())
```

### Filtro 1 *(0,5 pt)*

### Filtro 5 *(0,5 pt)*

### Filtro 6 *(0,5 pt)*
Selecione vendas onde **`Faixa_Ticket == "Alto"`** E **`Desconto_Pct == 0`** (sem desconto).  
Em seguida, agrupe por `Categoria` e calcule a **soma do `Total_Liquido`** de cada categoria nesse grupo.  
Imprima o resultado ordenado do maior para o menor.

```python
# SEU CÓDIGO AQUI - Filtro 6
resultado6 = df[(df["Faixa_Ticket"] == "Alto") & (df["Desconto_Pct"] == 0)]
soma_por_categoria = resultado6.groupby("Categoria")["Total_Liquido"].sum().sort_values(ascending=False)
print(soma_por_categoria)
print(f"Registros encontrados: {len(resultado6)}")
```

### Filtro 7 - Desafio *(0,5 pt)*
Encontre os **produtos** que atendem **simultaneamente** a todos os critérios abaixo:

- Categoria **não é** `"Alimentos"`
- Preço unitário **entre R$ 80 e R$ 300** (inclusive)
- Quantidade vendida **maior ou igual a 5**
